```python
split_index = int(len(model_df) * 0.8)

train_df = model_df.iloc[:split_index].copy()
test_df = model_df.iloc[split_index:].copy()

print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)

print("\nTraining period:")
print(train_df["date"].min(), "to", train_df["date"].max())

print("\nTesting period:")
print(test_df["date"].min(), "to", test_df["date"].max())
```

```python
X_train = train_df.drop(columns=["sales", "date"])
y_train = train_df["sales"]

X_test = test_df.drop(columns=["sales", "date"])
y_test = test_df["sales"]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

categorical_features = [
    "store",
    "is_state_holiday",
    "is_school_holiday",
    "is_special_day"
]

numerical_features = [
    col for col in X_train.columns
    if col not in categorical_features
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            SimpleImputer(strategy="median"),
            numerical_features
        ),
        (
            "cat",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                ("encoder", OneHotEncoder(handle_unknown="ignore"))
            ]),
            categorical_features
        )
    ]
)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ))
])

rf_model.fit(X_train, y_train)
```

```python
y_pred = rf_model.predict(X_test)

print("First 10 predictions:")
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

r2 = r2_score(y_test, y_pred)

print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

## Final Hold-Out Evaluation (test.csv vs truth.csv)

Everything above was validated on a chronological split *within* `train.csv`. `test.csv` + `truth.csv` is a genuinely separate hold-out set the model has never seen, covering **Dec 2023 – May 2024** — entirely after the training period ends (train.csv stops Nov 2023). This is the real test of whether the model generalizes into the future, not just onto unseen days within the same season.

```python
# Load the true hold-out files (never used in training/validation above)
test = pd.read_csv("test.csv")
truth = pd.read_csv("truth.csv")

print("test:", test.shape)
print("truth:", truth.shape)
```

```python
# Retrain the winning model (Gradient Boosting) on ALL leakage-free sales data.
# Use demand_df (5142 rows) rather than the 3226-row surplus subset, since
# predicting sales doesn't require ordered/unsold at all -- more training data.

demand_model_df = demand_df.copy()
demand_model_df["year"] = demand_model_df["date"].dt.year
demand_model_df["month"] = demand_model_df["date"].dt.month
demand_model_df["day"] = demand_model_df["date"].dt.day
demand_model_df["day_of_week"] = demand_model_df["date"].dt.dayofweek
demand_model_df["week_of_year"] = demand_model_df["date"].dt.isocalendar().week.astype(int)
demand_model_df["is_weekend"] = (demand_model_df["day_of_week"] >= 5).astype(int)

X_full = demand_model_df.drop(columns=["sales", "date"])
y_full = demand_model_df["sales"]

final_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", GradientBoostingRegressor(
        n_estimators=200, learning_rate=0.05, max_depth=3, random_state=42
    ))
])
final_model.fit(X_full, y_full)
print("Final model trained on", X_full.shape[0], "rows")
```

```python
# Prepare test.csv the same way and predict
test_final = test.copy()
test_final["date"] = pd.to_datetime(test_final["date"])
test_final["year"] = test_final["date"].dt.year
test_final["month"] = test_final["date"].dt.month
test_final["day"] = test_final["date"].dt.day
test_final["day_of_week"] = test_final["date"].dt.dayofweek
test_final["week_of_year"] = test_final["date"].dt.isocalendar().week.astype(int)
test_final["is_weekend"] = (test_final["day_of_week"] >= 5).astype(int)

X_final_test = test_final.drop(columns=["row_id", "date"])
test_final["predicted_sales"] = final_model.predict(X_final_test)

# Score against truth.csv
submission_check = test_final[["row_id", "predicted_sales"]].merge(truth, on="row_id")

final_mae = mean_absolute_error(submission_check["sales"], submission_check["predicted_sales"])
final_rmse = np.sqrt(mean_squared_error(submission_check["sales"], submission_check["predicted_sales"]))
final_r2 = r2_score(submission_check["sales"], submission_check["predicted_sales"])

print("FINAL HOLD-OUT PERFORMANCE (test.csv vs truth.csv)")
print("=" * 50)
print("MAE :", round(final_mae, 4))
print("RMSE:", round(final_rmse, 4))
print("R²  :", round(final_r2, 4))
```

```python
# Sanity check against naive baselines, since a negative R2 alone is easy to
# misread. Compare the model against "always predict the training mean".
baseline_pred = np.full(len(truth), train["sales"].mean())
baseline_mae = mean_absolute_error(truth["sales"], baseline_pred)
baseline_r2 = r2_score(truth["sales"], baseline_pred)

comparison_table = pd.DataFrame({
    "Approach": ["Predict training mean (baseline)", "Gradient Boosting model"],
    "MAE": [baseline_mae, final_mae],
    "R2": [baseline_r2, final_r2]
})
print(comparison_table)

print("\nTrain-period mean sales:", round(train["sales"].mean(), 3))
print("Test-period (truth) mean sales:", round(truth["sales"].mean(), 3))
```

**What this shows:** the model beats the naive baseline (MAE 0.766 vs 0.869, less negative R²), but both are well below the in-train-period performance (R² 0.77–0.82). The reason is a real **distribution shift**: the test period's average standardized sales value (≈ -0.38) is noticeably lower than the training period's (≈ 0.12). That's a genuine finding worth reporting, not a bug — the model captures store/weather/weekday patterns well within a season, but a ~6-month-ahead forecast into a new season is a harder problem, and the honest conclusion is that this model is reliable for near-term (days–weeks ahead) forecasting rather than long-range extrapolation.

### Follow-up investigation: is this a `year`-extrapolation problem, or something deeper?

Gradient Boosting is tree-based and cannot extrapolate beyond the range of values it was trained on. `year` is the 2nd most important feature in the model (after `day_of_week`), and `train.csv` only covers 2021–2023 — so a natural hypothesis is that the negative R² above is simply the model failing to handle years it has never seen (including any *real* current-date prediction, e.g. today's actual date). We test that hypothesis directly below rather than assume it.

```python
# Retrain WITHOUT the year feature at all, and re-check true hold-out performance.
demand_model_df_no_year = demand_df.copy()
demand_model_df_no_year["month"] = demand_model_df_no_year["date"].dt.month
demand_model_df_no_year["day"] = demand_model_df_no_year["date"].dt.day
demand_model_df_no_year["day_of_week"] = demand_model_df_no_year["date"].dt.dayofweek
demand_model_df_no_year["week_of_year"] = demand_model_df_no_year["date"].dt.isocalendar().week.astype(int)
demand_model_df_no_year["is_weekend"] = (demand_model_df_no_year["day_of_week"] >= 5).astype(int)

X_no_year = demand_model_df_no_year.drop(columns=["sales", "date"])
y_no_year = demand_model_df_no_year["sales"]

numerical_features_no_year = [c for c in X_no_year.columns if c not in categorical_features]
preprocessor_no_year = ColumnTransformer(transformers=[
    ("num", SimpleImputer(strategy="median"), numerical_features_no_year),
    ("cat", Pipeline([("imputer", SimpleImputer(strategy="most_frequent")),
                       ("encoder", OneHotEncoder(handle_unknown="ignore"))]), categorical_features)
])

model_no_year = Pipeline([
    ("preprocessor", preprocessor_no_year),
    ("model", GradientBoostingRegressor(n_estimators=200, learning_rate=0.05, max_depth=3, random_state=42))
])
model_no_year.fit(X_no_year, y_no_year)

test_no_year = test_final.drop(columns=["year"]).copy()
X_test_no_year = test_no_year.drop(columns=["row_id", "date", "predicted_sales"], errors="ignore")
pred_no_year = model_no_year.predict(X_test_no_year)

check_no_year = test_final[["row_id"]].copy()
check_no_year["predicted_sales"] = pred_no_year
check_no_year = check_no_year.merge(truth, on="row_id")

mae_no_year = mean_absolute_error(check_no_year["sales"], check_no_year["predicted_sales"])
rmse_no_year = np.sqrt(mean_squared_error(check_no_year["sales"], check_no_year["predicted_sales"]))
r2_no_year = r2_score(check_no_year["sales"], check_no_year["predicted_sales"])

print("WITHOUT year feature -- true hold-out (test.csv vs truth.csv)")
print("MAE :", round(mae_no_year, 4))
print("RMSE:", round(rmse_no_year, 4))
print("R²  :", round(r2_no_year, 4))
print()
print("For comparison, WITH year feature (from earlier):")
print("MAE :", round(final_mae, 4), " R²  :", round(final_r2, 4))
```

**Result: removing `year` made performance worse, not better** (R² dropped further, from about -0.19 to about -0.37). That rules out the extrapolation hypothesis as the main cause.

The real explanation is a **shift in the target distribution itself** between the training and test periods — not something any single input feature can fix:

```python
print("Training-period mean sales (standardized):", round(train["sales"].mean(), 4))
print("True future test-period mean sales (standardized):", round(truth["sales"].mean(), 4))
print()
print("Difference:", round(truth['sales'].mean() - train['sales'].mean(), 4),
      "-- roughly", round(abs(truth['sales'].mean() - train['sales'].mean()), 2), "standard deviations")
```

### Conclusion: reliability for real today/tomorrow predictions

**This model should not be presented as reliable for live, real-world today/tomorrow surplus predictions in its current form.** The evidence:

- True hold-out R² is negative regardless of whether `year` is included or excluded as a feature.
- The training and true-future periods have measurably different average demand — a genuine distribution shift, not a fixable modeling bug.
- Tree-based models cannot detect or correct for a shift in the target's overall level; only retraining on data from the *current* period can.

**What would make it production-viable:**
1. **Periodic retraining** on a rolling window of recent real sales data (e.g. weekly/monthly), so the model always reflects current demand levels rather than a static historical snapshot.
2. **Drift monitoring** — tracking live prediction error over time and triggering a retrain when it degrades.
3. Until such a pipeline exists, this model is appropriately scoped as: *reliable for near-term forecasting on dates similar in season/context to the training period (Jul 2022–Nov 2023), and a working demonstration of the forecasting-to-recommendation pipeline, but not validated for live today/tomorrow deployment.*

## Surplus Decision Logic

This is the actual FOODWISE output: compare what the model expects a store to sell against what was actually ordered, and flag the surplus risk so staff know where over-ordering is likely.

```python
# Use the surplus-aware subset (rows where ordered & unsold are known) to
# compare model-predicted demand against what was actually ordered.
surplus_eval = surplus_df.copy()
surplus_eval["year"] = surplus_eval["date"].dt.year
surplus_eval["month"] = surplus_eval["date"].dt.month
surplus_eval["day"] = surplus_eval["date"].dt.day
surplus_eval["day_of_week"] = surplus_eval["date"].dt.dayofweek
surplus_eval["week_of_year"] = surplus_eval["date"].dt.isocalendar().week.astype(int)
surplus_eval["is_weekend"] = (surplus_eval["day_of_week"] >= 5).astype(int)

X_surplus_eval = surplus_eval.drop(
    columns=["date", "sales", "unsold", "ordered", "potential_surplus", "surplus_rate", "balance_error"]
)
surplus_eval["predicted_demand"] = final_model.predict(X_surplus_eval)

# Predicted surplus = what was ordered minus what the model expects to sell
surplus_eval["predicted_surplus"] = surplus_eval["ordered"] - surplus_eval["predicted_demand"]

print("Correlation between predicted_surplus and actual unsold:",
      round(surplus_eval["predicted_surplus"].corr(surplus_eval["unsold"]), 3))
```

```python
def classify_surplus(x):
    if x < 0.10:
        return "Low"
    elif x < 0.30:
        return "Moderate"
    else:
        return "High"

surplus_eval["surplus_risk"] = surplus_eval["predicted_surplus"].apply(classify_surplus)

print(surplus_eval["surplus_risk"].value_counts())
print()
risk_by_store = (
    surplus_eval.groupby("store")["surplus_risk"]
    .value_counts(normalize=True)
    .unstack()
    .round(3)
)
risk_by_store
```

```python
plt.figure(figsize=(10, 6))

risk_by_store[["Low", "Moderate", "High"]].plot(
    kind="bar", stacked=True, ax=plt.gca(),
    color=["#4C9A6A", "#E8B84B", "#C0504D"]
)

plt.ylabel("Share of days")
plt.title("Surplus Risk Distribution by Store")
plt.legend(title="Risk level")
plt.tight_layout()
plt.savefig("surplus_risk_by_store.png", dpi=120)
plt.show()
```

**Reading this:** stores 5, 7, and 8 run High surplus risk on close to half their days — they are consistently over-ordering relative to what the model expects them to sell. Store 0 and store 3 look well-calibrated (mostly Low/Moderate). That's an actionable, store-specific recommendation: tighten ordering at stores 5/7/8 first, since that's where the savings are concentrated.

## Waste Cost Analysis (Dataset_Propely.csv)

**Note on scope:** this dataset (canteen meals, June–Aug 2025) uses a completely different schema and time period from the store sales data above (bakery stores, 2021–2023) — there's no shared key (no store/date overlap) to merge them on. It's treated here as an independent case study on where waste cost concentrates, rather than an extension of the sales model.

```python
waste = pd.read_csv("Dataset_Propely.csv")
waste["Date"] = pd.to_datetime(waste["Date"])

print("Total waste cost: $", round(waste["Cost_Loss"].sum(), 2))
print("Total waste weight:", round(waste["Waste_Weight_kg"].sum(), 2), "kg")

cost_by_food = waste.groupby("Food_Category")["Cost_Loss"].sum().sort_values(ascending=False)
cost_by_section = waste.groupby("Canteen_Section")["Cost_Loss"].sum().sort_values(ascending=False)
cost_by_meal = waste.groupby("Meal")["Cost_Loss"].sum().sort_values(ascending=False)

print("\nCost loss by food category:\n", cost_by_food.round(2))
print("\nCost loss by canteen section:\n", cost_by_section.round(2))
print("\nCost loss by meal:\n", cost_by_meal.round(2))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

cost_by_food.plot(kind="bar", ax=axes[0], color="#C0504D")
axes[0].set_title("Waste Cost by Food Category")
axes[0].set_ylabel("Cost Loss ($)")

cost_by_section.plot(kind="bar", ax=axes[1], color="#4C72B0")
axes[1].set_title("Waste Cost by Canteen Section")

cost_by_meal.plot(kind="bar", ax=axes[2], color="#55A868")
axes[2].set_title("Waste Cost by Meal")

plt.tight_layout()
plt.savefig("waste_cost_breakdown.png", dpi=120)
plt.show()
```

**Reading this:** Meat drives more than half of total waste cost (\$12,960 of \$24,006) despite being one of four categories — it's the highest-value, highest-risk item to over-prepare. Section C and meal type are far more evenly split, so category is the lever that matters most here, not section or meal time.

## Summary: Business Insights

1. **Sales forecasting works well in-season.** Gradient Boosting reaches R² ≈ 0.83 on near-term hold-out days, driven mainly by *store identity*, *day of week*, and *weekend flag* — operational rhythm matters more than weather.
2. **The model is not currently reliable for live today/tomorrow predictions.** True future hold-out R² is negative (≈ -0.19), and this persists even after removing the `year` feature (≈ -0.37 without it) — ruling out simple extrapolation as the cause. The real driver is a measured ~0.49 standard-deviation shift in average demand between the training and test periods. This is a genuine limitation of a static, one-time-trained model, not a bug to patch: it needs **periodic retraining on recent data** to stay accurate for real current-date use.
3. **Surplus is concentrated in specific stores.** Stores 5, 7, and 8 show High surplus risk on ~40–47% of days — ordering policy review should start there for the fastest payoff.
4. **Waste cost is concentrated in Meat**, at over half of total cost loss across the canteen dataset. Portion control or tighter meat-prep forecasting would have outsized impact relative to other categories.
5. **Recommended scope for this project's model**: reliable for near-term forecasting within/near the training window (Jul 2022–Nov 2023) and as a demonstration of the full forecasting-to-recommendation pipeline; not yet validated for live production deployment without a retraining loop.
